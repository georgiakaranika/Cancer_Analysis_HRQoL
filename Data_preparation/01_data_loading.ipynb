```python
df_2014 = pd.read_sas('../Data/data_raw/LLCP2014.XPT', format='xport')
df_2014.head()
```

```python
print(df_2014.shape)
print(df_2014.info())
```

```python
print('CNCRAGE' in df_2014.columns)
```

```python
n_cancer = df_2014['CNCRAGE'].notna().sum()
print(f"Cancer cases in 2014: {n_cancer}")
```

```python
df_2014_cancer = df_2014[df_2014['CNCRAGE'].notna()]
print(f"Total cancer cases in 2014: {len(df_2014_cancer)}")
```

```python
print(df_2014_cancer.shape)
```

```python
print(df_2014_cancer.columns.tolist())
```

```python
selected_columns = [
    # HRQoL
    'GENHLTH', 'PHYSHLTH', 'MENTHLTH', 'POORHLTH',
    
    # Cancer-specific
    'CNCRAGE', 'CNCRTYP1', 'CSRVTRT1', 'CSRVDOC1', 'CSRVPAIN',
    
    # Demographics
    '_AGE80', 'SEX', 'INCOME2', 'EDUCA', '_RACE',
    
    # Socioeconomic
    'HLTHPLN1', 'PERSDOC2', 'MEDCOST',
    
    # Comorbidities
    'DIABETE3', 'CVDINFR4', 'CVDCRHD4', 'CVDSTRK3', 'ADDEPEV2',
    
    # Behavioral
    'SMOKE100', 'EXERANY2', 'SLEPTIM1',
    
    # Temporal
    'IYEAR',
    
    # Geographic
    '_STATE',
    
    # Social support
    'MARITAL',
    
    # Healthcare utilization
    'CHECKUP1', 'DRVISITS'
]
```

```python
df_2014_selected = df_2014_cancer[selected_columns]
print(df_2014_selected.shape)
```

```python
print(df_2014_selected.isnull().sum())
```

```python
# Fillna(0)
df_2014_selected['POORHLTH'] = df_2014_selected['POORHLTH'].fillna(0)
df_2014_selected['DRVISITS'] = df_2014_selected['DRVISITS'].fillna(0)

# Fillna(-1) — Not applicable
df_2014_selected['CSRVDOC1'] = df_2014_selected['CSRVDOC1'].fillna(-1)
df_2014_selected['CSRVPAIN'] = df_2014_selected['CSRVPAIN'].fillna(-1)

# Drop rows with missing values
df_2014_selected = df_2014_selected.dropna(subset=['CNCRTYP1', 'CSRVTRT1', 'EXERANY2'])

# Verification
print(df_2014_selected.isnull().sum())
print(f"\nTotal records after cleaning: {len(df_2014_selected)}")
```

```python
duplicates = df_2014_selected.duplicated().sum()
print(f"Duplicate rows: {duplicates}")
```
