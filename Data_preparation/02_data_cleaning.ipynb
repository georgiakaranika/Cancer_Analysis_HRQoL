```python
# Rename mapping — standardize variable names across years
rename_map = {
    # Sex
    'SEXVAR': 'SEX',
    'SEX1': 'SEX',
    
    # Diabetes
    'DIABETE3': 'DIABETE',
    'DIABETE4': 'DIABETE',
    
    # Depression
    'ADDEPEV2': 'ADDEPEV',
    'ADDEPEV3': 'ADDEPEV',
    
    # Income
    'INCOME2': 'INCOME',
    'INCOME3': 'INCOME',
    
    # Personal doctor
    'PERSDOC2': 'PERSDOC',
    'PERSDOC3': 'PERSDOC',
    
    # Medical cost barrier
    'MEDCOST': 'MEDCOST_U',
    'MEDCOST1': 'MEDCOST_U',
    
    # Health insurance
    'HLTHPLN1': 'HLTHPLN',
    'PRIMINSR': 'HLTHPLN',
    'PRIMINS1': 'HLTHPLN',
    'PRIMINS2': 'HLTHPLN',
    
    # Cancer type
    'CNCRTYP1': 'CNCRTYP',
    'CNCRTYP2': 'CNCRTYP',
    
    # Cancer treatment
    'CSRVTRT1': 'CSRVTRT',
    'CSRVTRT2': 'CSRVTRT',
    'CSRVTRT3': 'CSRVTRT',
    
    # Race (2022 only)
    '_RACE1': '_RACE'
}
```

```python
selected_columns = [
    'GENHLTH', 'PHYSHLTH', 'MENTHLTH', 'POORHLTH',
    'CNCRAGE', 'CNCRTYP', 'CSRVTRT', 'CSRVDOC1', 'CSRVPAIN',
    '_AGE80', 'SEX', 'INCOME', 'EDUCA', '_RACE',
    'HLTHPLN', 'PERSDOC', 'MEDCOST_U',
    'DIABETE', 'CVDINFR4', 'CVDCRHD4', 'CVDSTRK3', 'ADDEPEV',
    'SMOKE100', 'EXERANY2', 'SLEPTIM1',
    'IYEAR', '_STATE', 'MARITAL', 'CHECKUP1', 'DRVISITS',
    '_LLCPWT'
]
```

```python
# Diagnosis for 2015 dataset
df_2015 = pd.read_sas('../Data/data_raw/LLCP2015.XPT', format='xport')

# 1. Check if CNCRAGE exists
print("1. Does CNCRAGE exist?", 'CNCRAGE' in df_2015.columns)

# 2. Check non-null values if it exists
if 'CNCRAGE' in df_2015.columns:
    print("2. Number of non-null values:", df_2015['CNCRAGE'].notna().sum())
    print("   Total rows in 2015 dataset:", len(df_2015))
else:
    # If not found, let's look for alternative cancer age variables
    cancer_cols = [c for c in df_2015.columns if 'CNCR' in c]
    print("   Alternative cancer-related columns found:", cancer_cols)
```

```python
all_years = []

for year in range(2014, 2025):
    # 2015 does not have the detailed cancer survivorship module (CNCRAGE)
    if year == 2015:
        print(f"2015: Skipped — Cancer module / CNCRAGE not available")
        continue
       
    filepath = f'../Data/data_raw/LLCP{year}.XPT'
    if not os.path.exists(filepath):
        print(f"{year}: File not found")
        continue
    
    print(f"Loading {year}...")
    df = pd.read_sas(filepath, format='xport')
    
    # Apply rename mapping
    df = df.rename(columns=rename_map)
    
    # Filter cancer cases only
    if 'CNCRAGE' not in df.columns:
        print(f"{year}: CNCRAGE not found — skipping")
        continue
    df = df[df['CNCRAGE'].notna()]

    
    # Select columns (keep only what exists)
    cols = [c for c in selected_columns if c in df.columns]
    df = df[cols]

    # Add year column
    df['YEAR'] = year
    
    
    all_years.append(df)
    print(f"{year}: {len(df)} cancer cases")

print("\nDone!")
```

```python
pre = [2014, 2016, 2017, 2018, 2019]
peak = [2020, 2021]
post = [2022, 2023, 2024]

for period, years in [('Pre-Pandemic', pre), ('Pandemic Peak', peak), ('Pandemic Recovery', post)]:
    total = sum(len(df) for df in all_years if df['YEAR'].iloc[0] in years)
    print(f"{period}: {total} cancer cases")
```

```python
for df in all_years:
    year = int(df['YEAR'].iloc[0])
    print(f"{year}: {len(df.columns)} columns → {sorted(df.columns.tolist())}")
```

```python
# Find the union of all columns
all_cols = set()
for df in all_years:
    all_cols.update(df.columns.tolist())

# Check which columns are missing per year
for df in all_years:
    year = int(df['YEAR'].iloc[0])
    missing = all_cols - set(df.columns.tolist())
    if missing:
        print(f"{year}: Missing → {sorted(missing)}")
    else:
        print(f"{year}: ✅ Complete")
```

```python
# Step 1: Ensure all DataFrames have the exact same columns and order before concatenation
for i, df in enumerate(all_years):
    # Add any missing columns from selected_columns as pd.NA (or NaN)
    for col in selected_columns:
        if col not in df.columns:
            df[col] = pd.NA
    
    # Reorder columns to match selected_columns exactly, plus the 'YEAR' column at the end
    all_years[i] = df[selected_columns + ['YEAR']]

# Step 2: Visual check — Verify that every single DataFrame now has identical columns
print("=== Visual Check of Harmonized Columns per Year ===")
for df in all_years:
    year = int(df['YEAR'].iloc[0])
    print(f"{year}: {len(df.columns)} columns → Matches selected_columns? {list(df.columns[:-1]) == selected_columns}")

# Step 3: Concatenate all yearly datasets into one master dataset
df_combined = pd.concat(all_years, ignore_index=True)

# Step 4: Final verification outputs
print("\n=== Master Dataset Summary ===")
print(f"Master dataset successfully created!")
print(f"Total shape (Rows, Columns): {df_combined.shape}")
print(f"Unique years included: {sorted(df_combined['YEAR'].unique())}")
```

```python
# Step 1: Check for duplicate rows in the combined master dataset
duplicates_count = df_combined.duplicated().sum()
print(f"Number of exact duplicate rows found: {duplicates_count}")

# If there are duplicates, drop them
if duplicates_count > 0:
    df_combined = df_combined.drop_duplicates().reset_index(drop=True)
    print(f"Duplicates removed. New shape: {df_combined.shape}")
else:
    print("No duplicate rows found. Dataset is clean.")

# Step 2: Check missing values per column to ensure structural missingness is as expected
print("\n=== Missing Values Summary (Top columns with missing data) ===")
missing_counts = df_combined.isnull().sum()
missing_pct = (missing_counts / len(df_combined)) * 100
missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Missing Percentage (%)': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])
```
